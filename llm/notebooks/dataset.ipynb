```python
import json
import pandas as pd
import torch
from datasets import Dataset, DatasetDict
from pathlib import Path

# Load the Dolly dataset
dolly_path = '/Users/lochoang/Documents/Workspaces/capstone-project/llm/data/databricks-dolly-15k/databricks_dolly15k_translated.json'

# Load the JSON file
try:
    with open(dolly_path, 'r', encoding='utf-8') as f:
        dolly_data = json.load(f)
    print(f"Successfully loaded data from {dolly_path}")
    print(f"Number of examples: {len(dolly_data)}")
    
    # Convert to DataFrame for easier inspection
    df = pd.DataFrame(dolly_data)
    
    # Display the first few rows
    print("\nDataset preview:")
    display(df.head())
    
    # Show column information
    print("\nColumns in the dataset:")
    for col in df.columns:
        print(f"- {col}")
    
    # Create a Hugging Face dataset
    dataset = Dataset.from_pandas(df)
    print("\nCreated Hugging Face dataset with shape:", dataset.shape)
    
except FileNotFoundError:
    print(f"Error: File not found at {dolly_path}")
except json.JSONDecodeError:
    print(f"Error: Could not parse JSON file at {dolly_path}")
except Exception as e:
    print(f"Error loading dataset: {e}")
```
